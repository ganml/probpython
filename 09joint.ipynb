```python
import numpy as np

# Number of accepted samples
N = 100_000
X = []
Y = []

while len(X) < N:
    # Proposal from Uniform(0,1) x Uniform(0,1)
    x_star = np.random.rand()
    y_star = np.random.rand()

    # Check if inside support 0 < x < y < 1
    if 0 < x_star < y_star < 1:
        # Acceptance probability = f(x,y) / M
        # Here f(x,y)=2 and M=2
        u = np.random.rand()
        if u <= 1:  # always true, kept for algorithm clarity
            X.append(x_star)
            Y.append(y_star)

X = np.array(X)
Y = np.array(Y)

# Event of interest
event = (X > 0) & (X < 0.5) & (Y > 0) & (Y < 1)

# Monte Carlo estimate
prob_estimate = np.mean(event)
print(f"P(0<X<0.5, 0<Y<1): {prob_estimate:.4f}")
```

```python
import numpy as np

# Define the joint support and probabilities
# (x, y, p)
joint_pmf = [
    (0, 0, 0.2),
    (0, 1, 0.3),
    (1, 0, 0.1),
    (1, 1, 0.4),
]

# Extract values for sampling
values = [(x, y) for x, y, _ in joint_pmf]
probs = [p for _, _, p in joint_pmf]

# Number of simulations
N = 100_000

# Sample indices according to joint pmf
indices = np.random.choice(len(values), size=N, p=probs)

# Generate samples
samples = np.array([values[i] for i in indices])
X = samples[:, 0]
Y = samples[:, 1]

# Monte Carlo estimate of E[XY]
E_XY_sim = np.mean(X * Y)
print(f"Estimated E[XY]: {E_XY_sim:.4f}")
```

```python
import numpy as np

# Number of accepted samples desired
N = 100_000

X = []
Y = []

while len(X) < N:
    # Proposal sample from g(x,y)
    x_star = np.random.rand()
    y_star = np.random.rand()

    # Uniform for acceptance
    u = np.random.rand()

    # Target density f(x,y) = 1 on (0,1)^2
    f_xy = 1.0
    g_xy = 1.0
    M = 1.0

    # Acceptance step
    if u <= f_xy / (M * g_xy):
        X.append(x_star)
        Y.append(y_star)

X = np.array(X)
Y = np.array(Y)

# Event of interest
event = (X > 0.1) & (X <= 0.5) & (Y > 0.2) & (Y <= 0.3)

# Probability estimate
prob_estimate = np.mean(event)
print("Estimated probability:", prob_estimate)
```

```python
import numpy as np

# Number of simulations
N = 100_000

# Simulate fair die rolls (values 1 through 6)
X = np.random.randint(1, 7, size=N)

# Define Y
Y = 7 - X

# Estimate covariance using definition
cov_sim = np.mean((X - np.mean(X)) * (Y - np.mean(Y)))
print(f"Estimated Cov(X, Y): {cov_sim:.4f}")
print(f"Exact Cov(X, Y): {-35/12:.4f}")
```

```python
import numpy as np
from collections import Counter

# number of simulations
N = 1_000_000

# simulate X1 and X2 (Bernoulli with p=0.5)
X1 = np.random.randint(0, 2, N)
X2 = np.random.randint(0, 2, N)

# compute order statistics
X_min = np.minimum(X1, X2)
X_max = np.maximum(X1, X2)

# count occurrences
pairs = list(zip(X_min, X_max))
counts = Counter(pairs)

# estimate probabilities
pmf = {k: v / N for k, v in counts.items()}

# display results
for key in sorted(pmf):
    print(f"P(X_(1)={key[0]}, X_(2)={key[1]}) = {pmf[key]:.4f}")
```
