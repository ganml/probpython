```python
import numpy as np
from scipy.stats import norm

# Parameters
mu = 100        # mean
sigma = 20      # standard deviation
alpha = 0.95    # confidence level
n = 10**5       # number of simulations

# Simulate losses
X = np.random.normal(mu, sigma, n)

# Empirical VaR (simulation)
VaR_sim = np.quantile(X, alpha)

# Theoretical VaR
VaR_theoretical = mu + sigma * norm.ppf(alpha)

print(f"Simulated VaR_{alpha}: {VaR_sim:.4f}")
print(f"Theoretical VaR_{alpha}: {VaR_theoretical:.4f}")
```

```python
import numpy as np

# Parameters
alpha = 0.75
n = 10**6

# Step 1: simulate (X, Y) uniformly over triangle
U1 = np.random.uniform(0, 1, n)
U2 = np.random.uniform(0, 1, n)

mask = (U1 + U2 > 1)
U1[mask] = 1 - U1[mask]
U2[mask] = 1 - U2[mask]

X = U1
Y = U2
S = X + Y

# Function to compute ES
def compute_es(data, alpha):
    var = np.quantile(data, alpha)
    es = data[data > var].mean()
    return var, es

# Compute results
VaR_X, ES_X = compute_es(X, alpha)
VaR_Y, ES_Y = compute_es(Y, alpha)
VaR_S, ES_S = compute_es(S, alpha)

print(f"ES_0.75(X): {ES_X:.4f}")
print(f"ES_0.75(Y): {ES_Y:.4f}")
print(f"ES_0.75(S): {ES_S:.4f}")
```
