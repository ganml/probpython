```python
import numpy as np

# Number of accepted samples we want
N = 100_000

accepted_x = []
accepted_y = []

# Rejection sampling from the triangle 0 < x < y < 1
while len(accepted_x) < N:
    x = np.random.rand()
    y = np.random.rand()

    if x < y:  # accept point in triangular region
        accepted_x.append(x)
        accepted_y.append(y)

X = np.array(accepted_x)
Y = np.array(accepted_y)

# Approximate conditioning window around Y = 0.8
eps = 0.005
mask = np.abs(Y - 0.8) < eps

# Conditional probability estimate
prob_est = np.mean(X[mask] < 0.2)

print(f"Estimated P(X < 0.2 | Y = 0.8): {prob_est:.4f}")
print(f"Number of conditional samples used: {mask.sum()}")
```

```python
import random

# Joint pmf as a list of (x, y, probability)
pmf = [(0, 1, 0.10),
    (1, 1, 0.15),
    (2, 1, 0.05),
    (0, 2, 0.20),
    (1, 2, 0.25),
    (2, 2, 0.25)]

# Build cumulative distribution for sampling
states = [(x, y) for x, y, _ in pmf]
probs = [p for _, _, p in pmf]

# Simulation
N = 100_000
count_y1 = 0
count_x1_y1 = 0

for _ in range(N):
    x, y = random.choices(states, weights=probs, k=1)[0]
    if y == 1:
        count_y1 += 1
        if x == 1:
            count_x1_y1 += 1

estimate = count_x1_y1 / count_y1
print(f"Estimated P(X=1 | Y=1): {estimate:.4f}")
```

```python
import numpy as np

# number of simulations
n = 1_000_000

# simulate X
X = np.random.choice([0, 1], size=n, p=[0.3, 0.7])

# conditional mean and std
mean_Y_given_X = 2 * X + 1
std_Y_given_X = 2  # since variance = 4

# simulate Y | X
Y = np.random.normal(loc=mean_Y_given_X, scale=std_Y_given_X)

# compute estimates
E_Y = np.mean(Y)
Var_Y = np.var(Y)

print(f"Estimated E[Y]: {E_Y:.4f}")
print(f"Estimated Var(Y): {Var_Y:.4f}")
```
