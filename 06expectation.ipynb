```python
import numpy as np

# parameters
N = 1_000_000   # number of simulations
k = 6           # moment order

# simulate X ~ Uniform(0,1)
X = np.random.uniform(0, 1, N)

# sample mean
mu_hat = np.mean(X)
# kth raw moment (simulation)
raw_moment_sim = np.mean(X**k)
# kth central moment (simulation)
central_moment_sim = np.mean((X - mu_hat)**k)
# theoretical values
raw_moment_theory = 1 / (k + 1)
if k % 2 == 1:
    central_moment_theory = 0.0
else:
    central_moment_theory = 1 / ((k + 1) * 2**k)

print(f"k = {k}")
print("Raw moment:")
print(f"  Simulation : {raw_moment_sim:.6f}")
print(f"  Theory     : {raw_moment_theory:.6f}")

print("\nCentral moment:")
print(f"  Simulation : {central_moment_sim:.6f}")
print(f"  Theory     : {central_moment_theory:.6f}")
```

```python
import numpy as np

# Define the values and probabilities
X_values = np.array([1, 3])
probabilities = np.array([0.5, 0.5])

# Number of samples for simulation
N = 1_000_000

# Simulate N samples from X
samples = np.random.choice(X_values, size=N, p=probabilities)

# Compute sample variance
var_sim = np.var(samples)

# Compute sample standard deviation
std_sim = np.std(samples)

print(f"Simulated variance of X: {var_sim:.6f}")
print(f"Simulated standard deviation of X: {std_sim:.6f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

t_vals = np.linspace(-5, 5, 200)
M_vals = (np.exp(t_vals) - 1) / t_vals
M_vals[t_vals == 0] = 1  # define M(0) by continuity

plt.figure(figsize=(4,3))
plt.plot(t_vals, M_vals, color="black")
plt.xlabel("t")
plt.ylabel(r"$M_X(t)$")
plt.title("MGF")
plt.savefig("mgf.pdf", bbox_inches='tight', pad_inches=0)
```
