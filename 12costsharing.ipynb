```python
import numpy as np

# Generate samples from f(x) = 2x on (0,1)
# Inverse CDF: F(x) = x^2 → X = sqrt(U)
def sample_X(n):
    U = np.random.uniform(0, 1, n)
    return np.sqrt(U)

def estimate_prob(d, num_sim=500000):
    X = sample_X(num_sim)
    Y = np.maximum(X - d, 0)
    return np.mean(Y < 0.5)

def find_d(target=0.64):
    low, high = 0.0, 1.0

    for _ in range(30):  # binary search
        mid = (low + high) / 2
        prob = estimate_prob(mid)

        # print progress if desired
        if _ % 5 == 0:
            print(f"d={mid:.4f}, prob={prob:.4f}")

        if prob < target:
            low = mid
        else:
            high = mid

    return (low + high) / 2

d_est = find_d()
print(f"Estimated deductible d: {d_est:.4f}")
```

```python
import numpy as np

# Sample from f(x) = 2x^{-3}, x > 1
def sample_X(n):
    U = np.random.uniform(0, 1, n)
    return (1 - U) ** (-0.5)

def estimate_expected_payment(num_sim=1_000_000, limit=10):
    np.random.seed(0)

    X = sample_X(num_sim)
    Y = np.minimum(X, limit)

    return np.mean(Y)

# Run simulation
expected_payment = estimate_expected_payment()
print(f"Estimated expected payment: {expected_payment:.4f}")
```

```python
import numpy as np

def simulate_variance_exponential(num_sim=100_000):
    # Set lambda so that Var(X) = 50
    lam = 1 / np.sqrt(50)

    # numpy uses scale = 1/lambda
    X = np.random.exponential(scale=1/lam, size=num_sim)

    # Apply 3% inflation
    Y = 1.03 * X

    # Estimate variance
    return np.var(Y)

estimated_var = simulate_variance_exponential()
print(f"Estimated variance: {estimated_var:.3f}")
```
