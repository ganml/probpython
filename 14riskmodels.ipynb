```python
import numpy as np

# Parameters
n_policies = 100
p = 0.01
claim_size = 10000
n_sim = 10**6

# Simulate claims: Binomial counts × claim size
# Each simulation: number of claims ~ Binomial(100, 0.01)
num_claims = np.random.binomial(n_policies, p, size=n_sim)

# Aggregate loss
S = num_claims * claim_size

# Estimate mean and variance
mean_S = np.mean(S)
var_S = np.var(S)

print(f"Estimated E[S]: {mean_S:.2f}")
print(f"Estimated Var(S): {var_S:.2f}")
```

```python
import numpy as np

# Parameters
lam = 5                # Poisson mean
n_sim = 10**6          # number of simulations

# Choose a severity distribution: Exponential(1)
# Then E[X^2] = 2
def sample_X(size):
    return np.random.exponential(scale=1.0, size=size)

# Simulate N
N = np.random.poisson(lam, size=n_sim)

# Simulate aggregate loss S
S = np.zeros(n_sim)

for i in range(n_sim):
    if N[i] > 0:
        S[i] = sample_X(N[i]).sum()

# Empirical variance
var_S_sim = np.var(S)

# Theoretical value: lambda * E[X^2]
EX2 = 2.0
var_S_theoretical = lam * EX2

print(f"Simulated Var(S): {var_S_sim:.4f}")
print(f"Theoretical Var(S): {var_S_theoretical:.4f}")
```

```python
import numpy as np

# Parameters
n_policies = 5
p = 0.1
loss = 1000
theta = 0.001
n_sim = 10**6

# Simulate portfolio losses
X = np.random.binomial(1, p, size=(n_sim, n_policies)) * loss
S = X.sum(axis=1)

# Estimate E[e^{theta S}]
exp_term = np.exp(theta * S)
mean_exp = np.mean(exp_term)

# Exponential premium
premium = (1 / theta) * np.log(mean_exp)

print(f"Exponential premium: {premium:.4f}")
```
