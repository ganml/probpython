```python
import numpy as np

# Number of simulations
n_sim = 100_000

# Simulate arrival times (in minutes after 10:00 a.m.)
np.random.seed(0)  # for reproducibility
arrival_times = np.random.uniform(0, 30, size=n_sim)

# Estimate probability of arriving within first 10 minutes
prob_estimate = np.mean(arrival_times <= 10)

print(f"Estimated probability: {prob_estimate:.4f}")
```

```python
import numpy as np

# Parameters
mu = 70      # mean height (inches)
sigma = 3    # standard deviation (inches)
n_sim = 100_000  # number of simulations

# Simulate heights
heights = np.random.normal(mu, sigma, n_sim)

# Estimate probability
prob_estimate = np.mean(heights > 75)

print(f"Estimated probability: {prob_estimate:.4f}")
```

```python
import numpy as np

# Parameters
lambda_rate = 2   # arrivals per unit time
n_sim = 1_000_000  # number of simulations

# Simulate interarrival times (Exponential distribution)
interarrival_times = np.random.exponential(1/lambda_rate, size=n_sim)

# Estimate probability that time between arrivals exceeds 1
prob_estimate = np.mean(interarrival_times > 1)

print(f"Estimated P(T > 1): {prob_estimate:.4f}")
```

```python
import numpy as np

# Number of simulations
n_sim = 100_000

# Simulate X ~ Gamma(shape=2, rate=2)
# numpy uses the scale parameter = 1 / rate
X = np.random.gamma(shape=2, scale=1/2, size=n_sim)

# Estimate mean and variance
mean_est = np.mean(X)
var_est = np.var(X)

print(f"Estimated mean: {mean_est:.4f}")
print(f"Estimated variance: {var_est:.4f}")
```

```python
import numpy as np

# Number of simulations
n_sim = 100_000

# Simulate X ~ Beta(6, 3)
X = np.random.beta(a=6, b=3, size=n_sim)

# Estimate P(X < 0.3)
prob_estimate = np.mean(X < 0.3)
prob_exact = 168*(0.3**6/6-2*0.3**7/7+0.3**8/8)

print(f"Estimated P(X < 0.3): {prob_estimate:.4f}")
print(f"Exact P(X < 0.3): {prob_exact:.4f}")
```

```python
import numpy as np

# Set parameters
mu = 2
sigma = 0.5
n_sim = 100_000  # number of simulations

# Simulate lognormal random variables
X = np.random.lognormal(mean=mu, sigma=sigma, size=n_sim)

# Estimate probability
prob_estimate = np.mean(X > 10)

print(f"Estimated P(X > 10): {prob_estimate:.4f}")
```
