```python
import numpy as np

# Parameter
lam = 0.01          # rate parameter (mean = 1/lambda)
n_sim = 10**6

np.random.seed(0)

# Simulate future lifetimes T_x ~ Exp(lambda)
T = np.random.exponential(scale=1/lam, size=n_sim)

# Curtate lifetime
K = np.floor(T)

# Estimate curtate expectation of life
e_x_sim = np.mean(K)
e_x_theo = 1/(np.exp(lam)-1)

print(f"Estimated curtate expectation e_x: {e_x_sim:.4f}")
print(f"Theoretical curtate expectation e_x: {e_x_theo:.4f}")
```

```python
import numpy as np

# Parameters
B = 0.00027
c = 1.1
x = 50
t = 10
n_sim = 10**6
dt = 0.01  # small time window

# Inverse survival function simulation for Gompertz
U = np.random.uniform(size=n_sim)

# cumulative hazard integral:
# H(t) = (B / ln(c)) * (c^(x+t) - c^x)

lnc = np.log(c)

# Solve for T using inversion:
# U = exp(-H(T)) => -ln(U) = H(T)

H_T = -np.log(U)

T = (np.log((H_T * lnc / B) + c**x) / lnc) - x

# Estimate density at t = 10
count = np.sum((T >= t) & (T < t + dt))
f_est = count / (n_sim * dt)

print(f"Estimated f_50(10): {f_est:.6f}")
```
