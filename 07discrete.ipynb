```python
import numpy as np
from math import comb

# Number of simulations
n_trials = 500_000
n_tosses = 10

# Simulate coin tosses (1 = head, 0 = tail)
tosses = np.random.binomial(1, 0.5, size=(n_trials, n_tosses))

# Count number of heads in each experiment
X = tosses.sum(axis=1)

# Simulated probability P(X = 4)
simulated_prob = np.mean(X == 4)

# Exact probability for comparison
exact_prob = comb(10, 4) * (0.5 ** 10)

print(f"Simulated P(X=4): {simulated_prob:.6f}")
print(f"Exact P(X=4): {exact_prob:.6f}")
```

```python
import numpy as np

# Set random seed for reproducibility
np.random.seed(0)

# Number of simulations
n_sim = 1_000_000

# Simulate Poisson(3) random variables
N = np.random.poisson(lam=3, size=n_sim)

# Estimate P(N < 3)
prob_estimate = np.mean(N < 3)

print(f"P(N<3): {prob_estimate:.6f}")
```

```python
import numpy as np

# Parameters
q = 0.4       # probability of success
r = 3         # number of successes
n_sim = 100_000  # number of simulations

failures = []
for _ in range(n_sim):
    success_count = 0
    failure_count = 0
    while success_count < r:
        if np.random.rand() < q:
            success_count += 1
        else:
            failure_count += 1
    failures.append(failure_count)

# Estimate probability P(N = 2)
prob_estimate = np.mean(np.array(failures) == 2)
print(f"Estimated P(N = 2): {prob_estimate:.4f}")
```

```python
import numpy as np
from math import comb

# Parameters
U = 50       # total bulbs
K = 12       # defective bulbs
n = 5        # sample size
target_k = 2 # number of defective bulbs we're interested in

# Number of simulations
n_sim = 100_000
# Simulation
count = 0

for _ in range(n_sim):
    # Draw n bulbs without replacement
    sample = np.random.choice([1]*K + [0]*(U-K), size=n, replace=False)
    # Count defective bulbs in the sample
    if np.sum(sample) == target_k:
        count += 1

# Estimate probability
prob_estimate = count / n_sim
# Exact probability for comparison
exact_prob = comb(12, 2) * comb(38, 3) / comb(50, 5)
print(f"Estimated P(X={target_k}): {prob_estimate:.6f}")
print(f"Exact P(X={target_k}): {exact_prob:.6f}")
```
