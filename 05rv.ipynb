```python
import numpy as np

# Parameters
p = 0.3          # probability of heads
n = 100000       # number of simulations

# Simulate coin tosses: H = 1, T = 0
coin_tosses = np.random.rand(n) < p

# Define Bernoulli random variable
X = coin_tosses.astype(int)

# Empirical probabilities
print(f"Estimated P(X = 1): {np.mean(X):.4f}")
print(f"Estimated P(X = 0): {1 - np.mean(X):.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
n_simulations = 100000  # number of trials

# Simulate tossing two fair coins: 0 = T, 1 = H
tosses = np.random.randint(0, 2, size=(n_simulations, 2))

# Define random variable X = number of heads
X = tosses.sum(axis=1)

# Estimate pmf
values, counts = np.unique(X, return_counts=True)
pmf = counts / n_simulations

# Plot pmf
plt.figure(figsize=(4,3))
plt.bar(values, pmf, color='gray')
plt.xlabel("Number of heads (X)")
plt.ylabel("Probability")
plt.title("Estimated pmf of X")
plt.savefig("pmf.pdf", bbox_inches='tight', pad_inches=0)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the pdf
x = np.linspace(0, 1, 500)
f = 2 * x

# Numerical integration to verify total probability
area = np.trapezoid(f, x)
print(f"Approximate integral of f(x) over [0,1]: {area:.4f}")

# Plot the pdf
plt.figure(figsize=(4,3))
plt.plot(x, f, color="gray")
plt.fill_between(x, f, alpha=0.3, color="gray")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("pdf")
plt.savefig("pdf.pdf", bbox_inches='tight', pad_inches=0)


```

```python
import numpy as np
import matplotlib.pyplot as plt

# Example: X takes values 1, 2, 3 with probabilities 0.2, 0.5, 0.3
values = np.array([1, 2, 3])
probabilities = np.array([0.2, 0.5, 0.3])

# Simulate a large number of trials
n_sim = 100000
samples = np.random.choice(values, size=n_sim, p=probabilities)

# Compute empirical CDF
x = np.linspace(0, 5, 500)
F_empirical = [np.mean(samples <= xi) for xi in x]

# Plot the CDF
plt.figure(figsize=(4,3))
plt.step(x, F_empirical, where='post', color="black")
plt.xlabel("x")
plt.ylabel("F_X(x)")
plt.title("Empirical cdf of X")
plt.grid(True)
plt.savefig("ecdf.pdf", bbox_inches='tight', pad_inches=0)
```

```python
import numpy as np
from scipy import stats

# Define the values and their probabilities
values = [1, 2, 3]
probabilities = [0.2, 0.5, 0.3]

# Simulate a large number of samples
np.random.seed(0)  # for reproducibility
samples = np.random.choice(values, size=100000, p=probabilities)

# Compute mode
mode = stats.mode(samples, keepdims=True).mode[0]

# Compute median
median = np.median(samples)

# Compute 75th percentile
percentile_75 = np.percentile(samples, 75)

print(f"Mode: {mode}")
print(f"Median: {median}")
print(f"75th percentile: {percentile_75}")
```
